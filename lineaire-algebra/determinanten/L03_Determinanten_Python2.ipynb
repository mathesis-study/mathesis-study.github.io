# Met Python

We controleren hetzelfde voorbeeld met Python. We schrijven de matrix als een lijst van rijen. Daarna berekenen we $ad-bc$ en gebruiken we de formule voor de inverse van een $2\times 2$-matrix.

```python
A = [[31, -4],
     [18, -2]]

a, b = A[0]
c, d = A[1]
determinant = a*d - b*c

if determinant == 0:
    print("De matrix heeft geen inverse.")
else:
    inverse = [[d/determinant, -b/determinant],
               [-c/determinant, a/determinant]]
    for rij in inverse:
        print(rij)
```

Je kan de getallen in `A` wijzigen om een andere inverse te controleren. Deze code geldt enkel voor matrices van orde $2$.
