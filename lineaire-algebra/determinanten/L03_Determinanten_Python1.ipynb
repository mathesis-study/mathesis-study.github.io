# Met Python

We controleren hetzelfde voorbeeld met Python. We schrijven de matrix als een lijst van rijen. Elke rij is zelf een lijst van drie getallen. Daarna geven we de negen elementen een naam en passen we de regel van Sarrus toe.

```python
A = [[3, -2, 1],
     [2,  1, 2],
     [3,  2, 4]]

a, b, c = A[0]
d, e, f = A[1]
g, h, i = A[2]

hoofddiagonalen = a*e*i + b*f*g + c*d*h
nevendiagonalen = c*e*g + a*f*h + b*d*i
print(hoofddiagonalen - nevendiagonalen)
```

Je kan de getallen in `A` wijzigen om een andere determinant te controleren. Deze code gebruikt de regel van Sarrus en geldt dus enkel voor matrices van orde $3$.
